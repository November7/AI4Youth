```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules


data = [
    {'Transaction': 1, 'Items': ['mleko', 'dżem', 'ser', 'chleb', 'masło']},
    {'Transaction': 2, 'Items': ['masło']},
    {'Transaction': 3, 'Items': ['jajka', 'ser', 'chleb', 'masło', 'dżem', 'mleko']},
    {'Transaction': 4, 'Items': ['jajka', 'mleko', 'ser']},
    {'Transaction': 5, 'Items': ['ser', 'jajka', 'masło', 'chleb', 'dżem', 'mleko']},
    {'Transaction': 6, 'Items': ['masło', 'ser']},
    {'Transaction': 7, 'Items': ['chleb', 'dżem', 'ser', 'masło', 'mleko', 'jajka']},
    {'Transaction': 8, 'Items': ['chleb', 'mleko', 'jajka', 'masło', 'ser']},
    {'Transaction': 9, 'Items': ['ser', 'chleb', 'dżem']},
    {'Transaction': 10, 'Items': ['dżem', 'ser', 'mleko', 'masło', 'chleb', 'jajka']},
    {'Transaction': 11, 'Items': ['ser', 'dżem', 'masło', 'mleko']},
    {'Transaction': 12, 'Items': ['chleb', 'mleko']},
    {'Transaction': 13, 'Items': ['jajka', 'masło', 'ser', 'mleko', 'chleb', 'dżem']},
    {'Transaction': 14, 'Items': ['ser', 'mleko', 'dżem', 'chleb']},
    {'Transaction': 15, 'Items': ['dżem', 'masło', 'ser', 'chleb', 'jajka']},
    {'Transaction': 16, 'Items': ['jajka', 'chleb', 'masło', 'ser']},
    {'Transaction': 17, 'Items': ['ser']},
    {'Transaction': 18, 'Items': ['ser', 'jajka']},
    {'Transaction': 19, 'Items': ['chleb', 'jajka', 'mleko']},
    {'Transaction': 20, 'Items': ['ser', 'jajka', 'mleko', 'chleb', 'masło', 'dżem']}
]


df = pd.DataFrame(data)
df
```

```python
kodowanie_bin = df['Items'].str.join('|').str.get_dummies().astype(bool)

czeste_powiazania = apriori(kodowanie_bin, min_support=0.5, use_colnames=True)

reguly = association_rules(czeste_powiazania, metric="confidence", min_threshold=0.7)

print(f"Frequent Itemsets: {czeste_powiazania}")

```
