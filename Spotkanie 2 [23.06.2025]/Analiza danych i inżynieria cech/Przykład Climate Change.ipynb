## Climate Change

Przykład na zbiorze: https://www.kaggle.com/datasets/berkeleyearth/climate-change-earth-surface-temperature-data?datasetId=29&searchQuery=seas&select=GlobalLandTemperaturesByCountry.csv

Import bibliotek

```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('GlobalLandTemperaturesByCountry.csv',sep=',', encoding='utf8',decimal='.')
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum() 
# df['Country'].unique()
```

```python
lista_panstw = ['Poland', 'Germany','Czech Republic','United Kingdom', 'France', 'Italy','Canada', 'United States', 'China', 'India', 'Japan', 'Australia', 'Brazil', 'Russia', 'South Africa']
wybrane_kraje = df[df['Country'].isin(lista_panstw)].copy().reset_index()
wybrane_kraje.head()
```

```python
wybrane_kraje["Date"] = pd.to_datetime(wybrane_kraje["dt"])
wybrane_kraje["Year"] = wybrane_kraje["Date"].dt.year
wybrane_kraje["Month"] = wybrane_kraje["Date"].dt.month
wybrane_kraje["Day"] = wybrane_kraje["Date"].dt.day
# wybrane_kraje.head()
wybrane_kraje.dropna(subset=['AverageTemperature'], inplace=True)
wybrane_kraje.drop(['dt'], inplace=True, axis=1)
wybrane_kraje.head()
```

```python
tmp = wybrane_kraje.groupby(['Year', 'Country'])['AverageTemperature'].mean().reset_index()
tmp = tmp.sort_values('AverageTemperature', ascending=True)
tmp = tmp.sort_values('Year', ascending=True)
tmp
```

## Wizualizacja globu za pomocą Plotly

```python
fig = px.choropleth(tmp,locations='Country', locationmode='country names', 
                    color='AverageTemperature', color_continuous_scale='temps',
                    hover_name='Country', title='Średnia temperatura w wybranych krajach',
                    animation_frame='Year', animation_group='Country',
                    height=600, width=1000)

fig.update_layout(title_x=0.5, title_y=0.95, title_font_size=20)

fig.update_geos(projection_type="orthographic", showocean=True, oceancolor="LightBlue",
                showland=True,showlakes=False, lakecolor="LightBlue")

fig.show()
```
