Przykład na zbiorze: https://www.kaggle.com/datasets/berkeleyearth/climate-change-earth-surface-temperature-data?datasetId=29&searchQuery=seas&select=GlobalLandTemperaturesByCountry.csv

Import bibliotek

```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('GlobalLandTemperaturesByCountry.csv',sep=',', encoding='utf8',decimal='.')
df.head()
```

```python
lista_panstw = ['Poland', 'Germany', 'United Kingdom', 'France', 'Italy','Canada', 'United States', 'China', 'India', 'Japan', 'Australia', 'Brazil', 'Russia', 'South Africa']
wybrane_kraje = df[df['Country'].isin(lista_panstw)]
wybrane_kraje.head()
```

```python
wybrane_kraje = wybrane_kraje.copy()
wybrane_kraje["Date"] = pd.to_datetime(wybrane_kraje["dt"])
wybrane_kraje["Rok"] = wybrane_kraje["Date"].dt.year
wybrane_kraje["Miesiac"] = wybrane_kraje["Date"].dt.month
wybrane_kraje["Dzien"] = wybrane_kraje["Date"].dt.day
# wybrane_kraje.head()
wybrane_kraje.dropna(subset=['AverageTemperature'], inplace=True)
```

```python
tmp = wybrane_kraje.groupby(['Country', 'Rok'])['AverageTemperature'].mean().reset_index()
tmp = tmp.sort_values('AverageTemperature', ascending=True)
tmp = tmp.sort_values('Rok', ascending=True)
```

```python
fig = px.choropleth(tmp,locations='Country',locationmode='country names', 
                    color='AverageTemperature',
                    hover_name='Country', title='Średnia temperatura w wybranych krajach',
                    animation_frame='Rok',animation_group='Country')

fig.update_geos(projection_type="orthographic", showocean=True, oceancolor="LightBlue",
                showland=True, landcolor="LightGreen", showlakes=True, lakecolor="LightBlue")
fig.update_layout(height = 600, width = 1000, title_x=0.5, margin={"r":0,"t":50,"l":0,"b":0})
fig.show()
```
