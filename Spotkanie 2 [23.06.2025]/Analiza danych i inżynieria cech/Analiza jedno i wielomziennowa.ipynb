Import bibliotek i generowanie sztucznych danych do przykładu

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Przykładowy zbiór danych
data = pd.DataFrame({
    'Wiek': [22, 35, 58, 45, 26, 39, 41, 30],
    'Płeć': ['K', 'M', 'M', 'K', 'K', 'M', 'M', 'K'],
    'Dochód': [3000, 5500, 7200, 4900, 3100, 5700, 6000, 4000],
    'Wydatki': [200, 450, 500, 300, 220, 420, 480, 310]
})
```

Analiza jednozmiennowa - histogram wieku badanych osób (niezależnie od płci)

```python
plt.figure(figsize=(10, 6))
sns.histplot(data['Wiek'], kde=True) #gładka krzywa gęstości rozkładu – tzw. Kernel Density Estimate (KDE).

plt.title('Rozkład wieku klientów')
plt.xlabel('Wiek')
plt.ylabel('Liczba klientów')
plt.show()
```

```python
sns.countplot(x='Płeć', data=data)
plt.title('Rozkład płci wśród klientów')
plt.xlabel('Płeć')
plt.ylabel('Liczba osób')
plt.show()
```

Analiza wielozmiennowa: zależność dochodu i wydatku (z podziałem na płeć)

```python
# Wielozmiennowa analiza: zależność dochodu i wydatków
plt.figure(figsize=(10, 6))
plt.grid(True)
sns.scatterplot(data=data, x='Dochód', y='Wydatki', hue='Płeć')
plt.title('Dochód vs Wydatki według płci')
plt.show()
```

```python
data['Płeć'] = data['Płeć'].map({'K': 0, 'M': 1})
corr_matrix = data[['Wiek', 'Dochód', 'Wydatki','Płeć']].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Macierz korelacji')
plt.show()
```
